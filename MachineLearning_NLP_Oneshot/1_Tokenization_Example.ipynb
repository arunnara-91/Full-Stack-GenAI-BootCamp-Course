```python
## diff between nltk and spacy for tokenization
# nltk is a library for natural language processing in Python. It provides tools for tokenization, 
# stemming, lemmatization, and more. NLTK is often used for educational purposes and research in NLP.
# spacy is a library for natural language processing in Python. 
# It provides tools for tokenization, part-of-speech tagging, named entity recognition, and more. Spacy is designed for production use and is optimized for speed and efficiency. Spacy is often used in industry for building NLP applications.
```

```python
# corpus is a collection of text data that can be used for natural language processing tasks.
# it can be a single document, a collection of documents, or even a large dataset of text data. also book, article, news, etc can be considered as corpus.
# document is a single piece of text data that can be used for natural language processing tasks. 
# it can be a sentence, a paragraph, or even a single word. also book, article, news, etc can be considered as document.
# paragraph is a collection of sentences that are grouped together to form a coherent piece of text.
# sentence is a collection of words that are grouped together to form a complete thought or idea.
```

```python
corpus="""I am Arun, I am a Data Engineer.
I am learning Natural Language Processing. Let's Keep learning and keep growing!.."""
corpus
```

```python
print(corpus)
```

```python
##Tokenization
###  paragraph to Sentence tokenization
from nltk.tokenize import sent_tokenize, word_tokenize
```

```python
sent_tokenize(corpus)
```

```python
documents = sent_tokenize(corpus)
# along with pullstop, it will consider !. as a sentence end, so it will split into 3 sentences instead of 2 sentences.
documents
```

```python
documents
type(documents)
```

```python
for sentence in documents:
    print(sentence)
```

```python
#word tokenization
# word tokenization is the process of breaking a text into individual words or tokens.
# word_tokenize is a function in the nltk library that takes a string of text as input and returns a list of individual words or tokens.
# paragraph -> word tokenization
# sentences -> word tokenization
# word_tokenize will split the sentences into words and also remove the punctuation.
```

```python
from nltk.tokenize import word_tokenize
```

```python
word_tokenize(corpus)
```

```python
for sentence in documents:
    print(word_tokenize(sentence))
```

```python
from nltk.tokenize import wordpunct_tokenize
# diff between word_tokenize and wordpunct_tokenize
# word_tokenize is a function in the nltk library that takes a string of text as input  and returns a list of individual words or tokens. it will split the sentences into words and also remove the punctuation.
# wordpunct_tokenize is a function in the nltk library that takes a string of text as input and returns a list of individual words or tokens. it will split the sentences into words but it will not remove the punctuation.
# It will be treated as a separate token/ seaparate word. so it will split the sentences into words and also keep the punctuation as a separate token.    
```

```python
wordpunct_tokenize(corpus)
```

```python
from nltk.tokenize import TreebankWordTokenizer
# it will consider the last full stop as a sentence end.
# mostly will use word_tokenize nad sent_tokenize for tokenization. but if we want to keep the punctuation as a separate token then we can use wordpunct_tokenize.
```

```python
tokenizer=TreebankWordTokenizer()
tokenizer.tokenize(corpus)
```
